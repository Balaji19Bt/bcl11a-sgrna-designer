```python
!pip install scanpy
import scanpy as sc

adata = sc.datasets.paul15()
print(adata)
```

```python
sc.pp.normalize_total(adata)
sc.pp.log1p(adata)
sc.pp.pca(adata)
print("Done! PCA computed.")
```

```python
sc.pp.neighbors(adata, n_neighbors=10, n_pcs=20)
sc.tl.draw_graph(adata)
print("Done! Neighbors and graph computed.")
```

```python
sc.pp.neighbors(adata, n_neighbors=10, n_pcs=20)
sc.tl.draw_graph(adata)
print("Done! Neighbors and graph computed.")
```

```python
sc.tl.diffmap(adata)
sc.pp.neighbors(adata, n_neighbors=10, use_rep='X_diffmap')
sc.tl.draw_graph(adata)
print("Done! Diffusion map computed.")
```

```python
sc.tl.dpt(adata)
print("Done! Pseudotime computed.")
```

```python
sc.pl.draw_graph(adata, color=['paul15_clusters', 'dpt_pseudotime'], legend_loc='on data')
```

```python
sc.tl.rank_genes_groups(adata, groupby='paul15_clusters', method='wilcoxon')
sc.pl.rank_genes_groups(adata, n_genes=10, sharey=False)
```
